## Comparison between Asynchronous and Synchronous Implementations

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

### Using existing synchronous implementation .csv files

```python
sync_path = '../comparison/csv/sync'
csv_files = [f for f in os.listdir(sync_path) if f.endswith('.csv')]
```

### All Sync Experiments (Test Accuracy vs Rounds)

```python
plt.figure(figsize=(9, 5))
for csv_file in csv_files:
    file_path = os.path.join(sync_path, csv_file)
    sync_df = pd.read_csv(file_path)
    # sync_df = sync_df.head(50)
    sample_alpha = sync_df['sample_alpha'].iloc[0]
    label_alpha = sync_df['label_alpha'].iloc[0]
    label = f'sample_alpha={sample_alpha}, label_alpha={label_alpha}'
    plt.plot(sync_df['round'], sync_df['test_accuracy'], marker='.', label=label)

plt.xlabel('Round')
plt.ylabel('Test Accuracy')
plt.title('Test Accuracy vs Rounds')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show() 
```

### All Sync Experiments (Test Accuracy vs Time)

```python
plt.figure(figsize=(9, 5))
for csv_file in csv_files:
    file_path = os.path.join(sync_path, csv_file)
    sync_df = pd.read_csv(file_path)
    sample_alpha = sync_df['sample_alpha'].iloc[0]
    label_alpha = sync_df['label_alpha'].iloc[0]
    label = f'sample_alpha={sample_alpha}, label_alpha={label_alpha}'
    plt.plot(sync_df['timedelta'], sync_df['test_accuracy'], marker='.', label=label)

plt.xlabel('Time')
plt.ylabel('Test Accuracy')
plt.title('Test Accuracy vs Rounds')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show() 
```

### Comparison between Asynchronous and Synchronous Implementations

Label Alpha = 0.1 | Sample Alpha = 1.0

```python
label_alpha = 0.1        # Change label alpha. Choose from [0.1, 1.0, 10.0]
sample_alpha = 1.0      # Change sample alpha. Choose from [1.0, 3.0, 30.0]

async_df = pd.read_csv(f"../comparison/csv/async/test_accuracy_L{label_alpha}_S{sample_alpha}.parquet")    # Change label and sample here
sync_df = pd.read_csv(f"../comparison/csv/sync/test_accuracy_L{label_alpha}_S{sample_alpha}.parquet")      # Change label and sample here

async_df = async_df.head(50)
sync_df = sync_df.head(50)

# Create figure with 4 subplots arranged in 2x2 grid
# fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(15, 10))

# Create figure with 3 plots arranged in 3x1 grid
fig, (ax1, ax2, ax3) = plt.subplots(3, 1, figsize=(10, 15))

# First plot: Test Accuracy vs. Rounds
ax1.plot(async_df['round'], async_df['test_accuracy'], label='Async', marker='o')
ax1.plot(sync_df['round'], sync_df['test_accuracy'], label='Sync', marker='s')
ax1.set_xlabel('Round')
ax1.set_ylabel('Test Accuracy')
ax1.set_title(f'Test Accuracy vs. Rounds (L={label_alpha}, S={sample_alpha})')
ax1.legend()
ax1.grid(True)

# Second plot: Test Accuracy vs. Time Delta
ax2.plot(async_df['timedelta'], async_df['test_accuracy'], label='Async', marker='o')
ax2.plot(sync_df['timedelta'], sync_df['test_accuracy'], label='Sync', marker='s')
ax2.set_xlabel('Time Delta')
ax2.set_ylabel('Test Accuracy')
ax2.set_title(f'Test Accuracy vs. Time Delta (L={label_alpha}, S={sample_alpha})')
ax2.legend()
ax2.grid(True)

# Third plot: Test Accuracy vs. Time
ax3.plot(async_df['time'], async_df['test_accuracy'], label='Async', marker='o')
ax3.plot(sync_df['time'], sync_df['test_accuracy'], label='Sync', marker='s')
ax3.set_xlabel('Time')
ax3.set_ylabel('Test Accuracy')
ax3.set_title(f'Test Accuracy vs. Time (L={label_alpha}, S={sample_alpha})')
ax3.legend()
ax3.grid(True)

ax3.set_xticks(ax3.get_xticks()[::40])  # Show every 10th tick, adjust as needed


# Fourth plot: hide it
#ax4.axis('off')  


plt.tight_layout()
plt.show()
```
